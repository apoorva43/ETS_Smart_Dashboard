## Quantile Dotplot

Instead of plotting the BRR distribution as a histogram or CI ribbon, represent it as a grid of `n` dots where each dot = one equally-likely outcome with probability = 1/n.

```python
# Baseline: Two countries, 2022, n_dots=20
fig = plot_quantile_dotplot(
    df,
    subject="MATH",
    countries=["CAN", "USA"],
    year=2022,
    n_dots=20,
    oecd_mean=472
)
plt.show()
```

```python
# n_dots=40, each=2.5%
fig = plot_quantile_dotplot(
    df,
    subject="MATH",
    countries=["CAN", "USA"],
    year=2022,
    n_dots=40,
    oecd_mean=472
)
plt.show()
```

```python
# n_dots=80, each=1.25%
fig = plot_quantile_dotplot(
    df,
    subject="MATH",
    countries=["CAN", "USA"],
    year=2022,
    n_dots=80,
    oecd_mean=472
)
plt.show()
```
